# Extra Plotly Charts

Extra charts for Plotly that mimic the Plotly Express interface

the new charts are:
- Waterfall
- Waffle
- Dumbbell
- Range
- Lollipop
- Slope
- Metric

Below are examples of each type with variations in the parameters that are set.

Import the required libraries

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import px_xtras2 as pxx
```

## Waterfall Chart - Profit Breakdown

The standard parameters are given, but with additional layout values

```python
wf_data = pd.DataFrame({
    "Category": ["Revenue", "COGS", "Operating Expenses", "Taxes", "Net Profit"],
    "Value": [100000, -50000, -20000, -10000, 0],
})
fig_wf = pxx.waterfall(df=wf_data, 
                   categories_col='Category',
                   values_col='Value',
                   title="Company Profit Waterfall",
                   height=600,
                   width=800,
                   template='plotly_dark'
                   )
fig_wf.show()
```

## Waterfall Chart - Profit/Loss

Default parameters give the following colour scheme

```python
waterfall_df = pd.DataFrame({
    'Category': ["Year beginning", "Profit1", "Loss1", "Q1", "Profit2", "Loss2", "Q2"],
    'Value': [100, 50, -20, 0, 40, -10, 0],
    'measure': ['absolute','relative', 'relative','total', 'relative','relative','total']

})
fig_waterfall = pxx.waterfall(df=waterfall_df, 
                             categories_col='Category', 
                             values_col='Value', 
                             measure_col='measure',
                             title="Profit/Loss",
                             )
fig_waterfall.show()
```

We can set the coñours of individual elements, or set them all to the same colour

```python
waterfall_df = pd.DataFrame({
    'Category': ["Year beginning", "Profit1", "Loss1", "Q1", "Profit2", "Loss2", "Q2"],
    'Value': [100, 50, -20, 0, 40, -10, 0],
    'measure': ['absolute','relative', 'relative','total', 'relative','relative','total']

})
fig_waterfall = pxx.waterfall(
                             categories=["Year beginning", "Profit1", "Loss1", "Q1", "Profit2", "Loss2", "Q2"], 
h                             data=[100, 50, -20, 0, 40, -10, 0], 
                             measure=['absolute','relative', 'relative','total', 'relative','relative','total'],
                             title="Profit/Loss- list",
                             color='blue', 
                             )
fig_waterfall.show()
```

## Waffle Chart - Example 1

Here we accept the default colours, which are a range of lightblue to blue

```python
waffle_data = {'Others': 54, 'LD': 72, 'Conservative': 121, 'Labour': 403}
waffle_df = pd.DataFrame(list(waffle_data.items()), columns=['Party', 'Seats'])

fig_waffle = pxx.waffle(df=waffle_df, 
                       categories_col='Party', 
                       values_col='Seats',
                       title='Waffle Chart of MPs',
                        width=600,
                        template='plotly_white'
                       )
fig_waffle.show()
```

## Waffle Chart - Example 2

Here we set a range of colours that are meaningful to the data (i.e the party colours)

```python
fig_waffle = pxx.waffle(df=waffle_df, 
                       categories_col='Party', 
                       values_col='Seats',
                       title='Waffle Chart of MPs (Custom Colors)',
                       local_colors=['grey', 'yellow', 'blue', 'red'],
                       width=600,
                       template='plotly_dark'
                       )
fig_waffle.show()
```

## Waffle Chart - Example 3

Here we set  range of colours

```python
fig_waffle = pxx.waffle(df=waffle_df, 
                       categories_col='Party', 
                       values_col='Seats',
                       title='Waffle Chart of MPs (Custom Width)',
                       startcolor='yellow',
                       endcolor='darkgreen',
                        template='plotly_dark',
                        width = 800
                       )
fig_waffle.show()
```

## Dumbbell Chart - Horizontal

```python
dumbbell_df = pd.DataFrame({
    'Country': ['USA', 'Canada', 'UK', 'Germany', 'France'],
    '2020': [10, 12, 9, 14, 11],
    '2021': [12, 15, 11, 15, 12]
})

fig_dumbbell = pxx.dumbbell(df=dumbbell_df, 
                           categories_col='Country', 
                           values1_col='2020', 
                           values2_col='2021',
                           label1='2020',
                           label2='2021',
                           title='Dumbbell Chart Example',
                           width=800,
                           height=600,
                           template='plotly_dark')
fig_dumbbell.show()
```

## Dumbbell Chart - Vertical

```python
fig_dumbbell_v = pxx.dumbbell(df=dumbbell_df, 
                             categories_col='Country', 
                             values1_col='2020', 
                             values2_col='2021',
                             label1='2020',
                             label2='2021',
                             title='Vertical Dumbbell Chart Example',
                             orientation='v',
                             width=500)
fig_dumbbell_v.show()
```

## Range Chart - Example 1

```python
fig_range = pxx.line_range(df=dumbbell_df, 
                           categories_col='Country', 
                           values1_col='2020', 
                           values2_col='2021',
                           label1='2020',
                           label2='2021',
                           title='Range Chart Example')
fig_range.show()
```

## Life Expectancy Data

Let's use some real data

```python
df_life = pd.read_csv('data/lifeExp.csv').round(1)
df_life
```

## Dumbbell Chart - Life Expectancy (Horizontal)

```python
fig_dumbbell_life = pxx.dumbbell(df=df_life, 
                           categories_col='country', 
                           values1_col='1952', 
                           values2_col='2002',
                           label1='1952',
                           label2='2002',
                           template="plotly_white",
                           height=400,
                           title='Life expectancy chart',
                           )
fig_dumbbell_life.show()
```

## Dumbbell Chart - Life Expectancy (Vertical)

```python
fig_dumbbell_life_v = pxx.dumbbell(df=df_life, 
                           categories_col='country', 
                           values1_col='1952', 
                           values2_col='2002',
                           label1='1952',
                           label2='2002',
                           title='Life expectancy chart (Vertical)',
                           template="none",
                           orientation='v'
                           )
fig_dumbbell_life_v.show()
```

## Range Chart - Life Expectancy

```python
fig_range_life = pxx.line_range(df=df_life, 
                           categories_col='country', 
                           values1_col='1952', 
                           values2_col='2002',
                           label1='1952',
                           label2='2002',
                           template="plotly_dark",
                           height=600,
                           title='Life Expectancy Range Chart',
                           width=800)
fig_range_life.show()
```

```python
lollipop_df = pd.DataFrame({
    'Category': ['A', 'B', 'C', 'D', 'E'],
    'Value': [10, 25, 15, 30, 20]
})
fig_lollipop = pxx.lollipop_chart(df=lollipop_df, 
                                 categories_col='Category', 
                                 values_col='Value', 
                                 title='Lollipop Chart Example (Custom Height, Width and template)', 
                                 template="plotly_dark",
                                 height=600,
                                 width=800)
fig_lollipop.show()
```

## Metric

```python
fig_metric = pxx.metric(title="Sales", value=12345, delta=123, prefix="$")
fig_metric.show()
```

## Slope Chart

```python
slope_df = pd.DataFrame({
    'Year': [2020, 2021],
    'USA': [10, 12],
    'Canada': [12, 15],
    'UK': [9, 11],
    'Germany': [14, 15],
    'France': [11, 12],
})
fig_slope = pxx.slope_chart(df=slope_df, categories_col='Year', values_cols=['USA', 'Canada', 'UK', 'Germany', 'France'], 
                            title='Slope Chart Example',
                            template='plotly_dark',
                            #font_size=20,
                            width=800,
                            height=600)
fig_slope.show()
```
